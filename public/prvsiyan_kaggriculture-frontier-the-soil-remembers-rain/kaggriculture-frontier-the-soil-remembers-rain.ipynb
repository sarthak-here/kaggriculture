## Why this composition survived

The modal route by itself was not publishable: three severe regressions appeared immediately in the consumed hard screen. Trace instrumentation showed the first differences were early weeds blocking route actions. This same gate discipline had already closed several unrelated historical ideas:

- a universal Ray replacement rescued many losses but created regressions and severe deltas;
- price-order sorting improved aggregate wins but turned a Moon win into a loss;
- visible-WEED repair passed fit, then failed its predeclared validation-gain threshold;
- the sheep/WOOL near-copy patch created seven win regressions and thirteen new severe tails;
- product-first, fifth-HIRE relocation, and late continuation variants were rejected earlier.

## Live execution proof

Frozen matrices are necessary but insufficient. The exact emitted `main.py` now plays a complete official-engine episode from both seats. This catches entry-point, reset, serialization, and environment-interface defects before packaging.


```python
from kaggle_environments import make

live_rows = []
for side in (0, 1):
    agents = [str(MAIN_PATH), "starter"] if side == 0 else ["starter", str(MAIN_PATH)]
    env = make("kaggriculture", configuration={"seed": 22750 + side}, debug=True)
    env.run(agents)
    statuses = [a.status for a in env.steps[-1]]
    rewards = [float(a.reward) for a in env.steps[-1]]
    assert statuses == ["DONE", "DONE"]
    live_rows.append({"candidate seat": side, "status": statuses[side], "reward": rewards[side], "opponent reward": rewards[1-side]})

live = pd.DataFrame(live_rows)
display(live)
assert set(live["status"]) == {"DONE"}
```

## Package integrity

The bundle contains exactly one root-level `main.py`. The notebook reopens it and verifies the digest of the packed bytes. The tarball is only a reproducibility artifact; Kaggle executes the notebook-emitted `main.py`.


```python
BUNDLE_PATH = Path("submission.tar.gz")
with tarfile.open(BUNDLE_PATH, "w:gz") as bundle:
    bundle.add(MAIN_PATH, arcname="main.py")

with tarfile.open(BUNDLE_PATH, "r:gz") as bundle:
    names = bundle.getnames()
    packed = bundle.extractfile("main.py").read()

assert names == ["main.py"]
assert hashlib.sha256(packed).hexdigest() == EXPECTED_AGENT_SHA256
print("bundle members", names)
print("bundle sha256", hashlib.sha256(BUNDLE_PATH.read_bytes()).hexdigest())
```
